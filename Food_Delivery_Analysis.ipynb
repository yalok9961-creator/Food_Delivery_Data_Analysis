```python
# Creating Food Delivery Dataset with 2000 Records

import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta

random.seed(42)
np.random.seed(42)

n = 2000

restaurants = [
    "Food Palace", "Pizza Hub", "Burger House",
    "Spice Garden", "Biryani Express",
    "Tasty Corner", "Cafe Delight", "Food Junction",
    "Urban Kitchen", "Royal Restaurant"
]

cuisines = [
    "Indian", "Chinese", "Italian",
    "Mexican", "Fast Food", "South Indian"
]

cities = [
    "Mumbai", "Thane", "Pune",
    "Nashik", "Nagpur", "Delhi"
]

payment_methods = [
    "UPI", "Credit Card", "Debit Card", "Cash"
]

statuses = ["Delivered", "Cancelled", "Pending"]

data = {
    "Order_ID": range(1001, 1001 + n),
    "Customer_ID": np.random.randint(101, 501, n),
    "Restaurant_Name": np.random.choice(restaurants, n),
    "Cuisine_Type": np.random.choice(cuisines, n),
    "City": np.random.choice(cities, n),
    "Order_Amount": np.random.randint(100, 1501, n),
    "Delivery_Time": np.random.randint(15, 61, n),
    "Payment_Method": np.random.choice(payment_methods, n),
    "Customer_Rating": np.random.randint(1, 6, n),
    "Order_Status": np.random.choice(statuses, n)
}

df = pd.DataFrame(data)

# Generate Order Dates
start_date = datetime(2026, 1, 1)

df["Order_Date"] = [
    start_date + timedelta(days=random.randint(0, 300))
    for _ in range(n)
]

# Generate Delivery Fees
df["Delivery_Fee"] = np.random.randint(20, 81, n)

# Calculate Total Amount
df["Total_Amount"] = (
    df["Order_Amount"] + df["Delivery_Fee"]
)

# Display first 5 records
df.head()
```

```python
print("Total Records:", len(df))
print("Total Columns:", len(df.columns))

print("\nDataset Shape:")
print(df.shape)
```

```python
# Save CSV File
df.to_csv("food_delivery_dataset.csv", index=False)

# Save Excel File
df.to_excel("food_delivery_dataset.xlsx", index=False)

print("Both dataset files saved successfully!")
```

```python
# Remove Duplicate Records

df.drop_duplicates(inplace=True)

print("Duplicate records removed successfully!")
print("Total Records:", len(df))
```

```python
# Check and Handle Missing Values

print("Missing Values Before Cleaning:")
print(df.isnull().sum())

# Fill missing numerical values
for col in df.select_dtypes(include=np.number).columns:
    df[col] = df[col].fillna(df[col].median())

# Fill missing categorical values
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].fillna(df[col].mode()[0])

print("\nMissing Values After Cleaning:")
print(df.isnull().sum())
```

```python
# Convert Order_Date into Datetime Format

df["Order_Date"] = pd.to_datetime(df["Order_Date"])

print("Date column converted successfully!")
print(df["Order_Date"].dtype)
```

```python
# Check Data Types

print(df.dtypes)
```

```python
# Create Additional Columns

df["Order_Month"] = df["Order_Date"].dt.month_name()

df["Order_Day"] = df["Order_Date"].dt.day_name()

df["Delivery_Category"] = np.where(
    df["Delivery_Time"] <= 30,
    "Fast Delivery",
    "Slow Delivery"
)

print("New columns created successfully!")

df.head()
```

```python
# Display Cleaned Dataset Information

print("Final Dataset Shape:", df.shape)
print("\nFinal Dataset:")
display(df.head(10))

print("\nMissing Values:")
print(df.isnull().sum().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())
```

```python
# Food Delivery Revenue Analysis

# Total Number of Orders
total_orders = len(df)

# Delivered Orders
delivered_orders = df[df["Order_Status"] == "Delivered"]

# Total Revenue from Delivered Orders
total_revenue = delivered_orders["Total_Amount"].sum()

# Average Order Value
average_order_value = delivered_orders["Total_Amount"].mean()

# Total Cancelled Orders
cancelled_orders = len(df[df["Order_Status"] == "Cancelled"])

# Total Pending Orders
pending_orders = len(df[df["Order_Status"] == "Pending"])

print("FOOD DELIVERY REVENUE ANALYSIS")
print("--------------------------------")
print("Total Orders:", total_orders)
print("Delivered Orders:", len(delivered_orders))
print("Cancelled Orders:", cancelled_orders)
print("Pending Orders:", pending_orders)
print("Total Revenue: ₹", round(total_revenue, 2))
print("Average Order Value: ₹", round(average_order_value, 2))
```

```python
# Restaurant-wise Revenue

restaurant_revenue = (
    delivered_orders.groupby("Restaurant_Name")["Total_Amount"]
    .sum()
    .sort_values(ascending=False)
)

print("Restaurant-wise Revenue:")
display(restaurant_revenue)
```

```python
# Top 5 Restaurants by Revenue

top_restaurants = restaurant_revenue.head(5)

print("Top 5 Restaurants by Revenue:")
display(top_restaurants)
```

```python
# Cuisine-wise Order Analysis

cuisine_orders = df["Cuisine_Type"].value_counts()

print("Cuisine-wise Orders:")
display(cuisine_orders)
```

```python
# Revenue by Cuisine Type

cuisine_revenue = (
    delivered_orders.groupby("Cuisine_Type")["Total_Amount"]
    .sum()
    .sort_values(ascending=False)
)

print("Cuisine-wise Revenue:")
display(cuisine_revenue)
```

```python
# City-wise Orders

city_orders = df["City"].value_counts()

print("City-wise Orders:")
display(city_orders)
```

```python
# Most Popular Payment Methods

payment_analysis = df["Payment_Method"].value_counts()

print("Payment Method Analysis:")
display(payment_analysis)
```

```python
# Delivery Time Analysis

average_delivery_time = delivered_orders["Delivery_Time"].mean()
fastest_delivery = delivered_orders["Delivery_Time"].min()
slowest_delivery = delivered_orders["Delivery_Time"].max()

print("Average Delivery Time:", round(average_delivery_time, 2), "minutes")
print("Fastest Delivery:", fastest_delivery, "minutes")
print("Slowest Delivery:", slowest_delivery, "minutes")
```

```python
# Customer Rating Analysis

average_rating = delivered_orders["Customer_Rating"].mean()

highest_rating = delivered_orders["Customer_Rating"].max()

lowest_rating = delivered_orders["Customer_Rating"].min()

print("CUSTOMER RATING ANALYSIS")
print("-------------------------")
print("Average Rating:", round(average_rating, 2))
print("Highest Rating:", highest_rating)
print("Lowest Rating:", lowest_rating)
```

```python
# Restaurant-wise Customer Ratings

restaurant_ratings = (
    delivered_orders.groupby("Restaurant_Name")["Customer_Rating"]
    .mean()
    .sort_values(ascending=False)
)

print("Restaurant-wise Average Ratings:")
display(restaurant_ratings)
```

```python
# Orders Above ₹500

high_value_orders = df[df["Total_Amount"] > 500]

print("High Value Orders:")
display(high_value_orders.head(10))

print("Total High Value Orders:", len(high_value_orders))
```

```python
# Orders Delivered Within 30 Minutes

fast_deliveries = delivered_orders[
    delivered_orders["Delivery_Time"] <= 30
]

print("Fast Delivery Orders:")
display(fast_deliveries.head(10))

print("Total Fast Deliveries:", len(fast_deliveries))
```

```python
# Orders with Rating Above 4

high_rated_orders = delivered_orders[
    delivered_orders["Customer_Rating"] > 4
]

print("Highly Rated Orders:")
display(high_rated_orders.head(10))

print("Total Highly Rated Orders:", len(high_rated_orders))
```

```python
# City-wise Revenue Analysis

city_revenue = (
    delivered_orders.groupby("City")["Total_Amount"]
    .sum()
    .sort_values(ascending=False)
)

print("City-wise Revenue:")
display(city_revenue)
```

```python
# Monthly Order Analysis

monthly_orders = df["Order_Month"].value_counts()

print("Monthly Orders:")
display(monthly_orders)
```

```python
# Revenue by Cuisine Type

plt.figure(figsize=(10, 6))

cuisine_revenue.plot(
    kind="bar",
    edgecolor="black"
)

plt.title("Revenue by Cuisine Type")
plt.xlabel("Cuisine Type")
plt.ylabel("Total Revenue (₹)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Top 5 Restaurants

plt.figure(figsize=(10, 6))

top_restaurants.plot(
    kind="bar",
    edgecolor="black"
)

plt.title("Top 5 Restaurants by Revenue")
plt.xlabel("Restaurant Name")
plt.ylabel("Revenue (₹)")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
# Orders by City

plt.figure(figsize=(9, 6))

city_orders.plot(
    kind="bar",
    edgecolor="black"
)

plt.title("Food Delivery Orders by City")
plt.xlabel("City")
plt.ylabel("Number of Orders")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
# Average Delivery Time by Restaurant

delivery_analysis = (
    delivered_orders.groupby("Restaurant_Name")["Delivery_Time"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))

delivery_analysis.plot(
    kind="bar",
    edgecolor="black"
)

plt.title("Average Delivery Time by Restaurant")
plt.xlabel("Restaurant Name")
plt.ylabel("Delivery Time (Minutes)")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
# Customer Rating Distribution

rating_counts = delivered_orders["Customer_Rating"].value_counts().sort_index()

plt.figure(figsize=(8, 5))

rating_counts.plot(
    kind="bar",
    edgecolor="black"
)

plt.title("Customer Rating Distribution")
plt.xlabel("Customer Rating")
plt.ylabel("Number of Orders")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Save Revenue by Cuisine Graph

plt.figure(figsize=(10, 6))
cuisine_revenue.plot(kind="bar")
plt.title("Revenue by Cuisine Type")
plt.xlabel("Cuisine Type")
plt.ylabel("Revenue (₹)")
plt.tight_layout()
plt.savefig("revenue_by_cuisine.png")
plt.show()
```

```python
# Export Filtered Orders

high_value_orders.to_csv(
    "filtered_orders.csv",
    index=False
)

print("Filtered orders CSV saved successfully!")
```

```python
# Create Excel Analysis Report

with pd.ExcelWriter("food_delivery_analysis.xlsx") as writer:

    df.to_excel(writer, sheet_name="All Orders", index=False)

    restaurant_revenue.to_frame(
        name="Total Revenue"
    ).to_excel(writer, sheet_name="Restaurant Revenue")

    cuisine_revenue.to_frame(
        name="Total Revenue"
    ).to_excel(writer, sheet_name="Cuisine Revenue")

    city_revenue.to_frame(
        name="Total Revenue"
    ).to_excel(writer, sheet_name="City Revenue")

    restaurant_ratings.to_frame(
        name="Average Rating"
    ).to_excel(writer, sheet_name="Restaurant Ratings")

print("Excel Analysis Report Created Successfully!")
```

```python
# Save Final Cleaned Dataset

df.to_csv("food_delivery_dataset.csv", index=False)

df.to_excel("food_delivery_dataset.xlsx", index=False)

print("Final datasets saved successfully!")
```
